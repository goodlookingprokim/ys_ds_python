## 제보 1. 오전·점심 대기 시간 열 개를 정리하다

> **이야기 위치:** 1막 · 현장 자료 입수
>
> **확보한 단서:** 오전·점심에 확인한 대기 시간 메모 10장이 도착했습니다.
>
> **이번 취재 임무:** 10개 기록을 2행 5열로 놓아 비교할 수 있는 모양으로 만듭니다.
>
> **다음 연결:** 모양을 갖추면 어느 구역에 사람이 몰렸는지 합계와 평균을 계산합니다.

첫 번째 현장 자료가 도착했습니다. 방송부는 축제 부스 다섯 곳의 줄 서는 시간을 오전과 점심에 한 번씩 확인했습니다. 오전 기록 5개와 점심 기록 5개를 합치면 모두 10개입니다. 실제 기록을 다루기 전에 연습하기 쉽게 `np.arange(10, 60, 5)`로 10분부터 55분까지 5분씩 늘어나는 값을 만듭니다.

### 배열은 언제 필요할까요?

대기 시간처럼 **같은 뜻과 단위를 가진 숫자**가 여러 개일 때 배열로 모으면 한꺼번에 계산하고 모양을 바꾸기 쉽습니다. 여기서는 부스 다섯 곳을 오전과 점심에 확인한 대기 시간 10개가 한 묶음입니다.

### 왜 여기서 배열을 쓸까요?

운영본부는 오전과 점심의 줄을 비교하려 합니다. 메모 10장을 배열에 순서대로 담으면 기록을 잃지 않은 채 오전 5개와 점심 5개로 나누고, 전체 합계도 한 번에 계산할 수 있습니다.

### 메모를 그대로 두면 왜 불편할까요?

숫자가 흩어져 있으면 어느 다섯 개가 오전인지 구분하기 어렵고, 합계를 낼 때 빠뜨리기 쉽습니다. 배열은 같은 종류의 값을 한 상자에 모아 순서와 개수를 컴퓨터가 함께 기억하게 합니다.

### arange와 reshape는 무엇을 바꿀까요?

`np.arange(10, 60, 5)`는 연습용 값 10부터 55까지를 5분 간격으로 만듭니다. 끝값 60은 포함하지 않습니다. `reshape(2, 5)`는 그 10개를 오전 윗줄과 점심 아랫줄에 5개씩 다시 놓을 뿐입니다. 숫자의 개수와 합계 **325분**은 바뀌지 않습니다.

> **장면 떠올리기:** 대기 시간 메모 10장을 책상 위에서 두 줄로 다시 놓아도 메모에 적힌 값은 그대로입니다.

**실행 전 예상:** 가장 짧은 값과 가장 긴 값은 몇 분이며, 두 줄로 바꾼 뒤 합계는 얼마일까요?


  ## 1-1. 간단한 NumPy 사용법

```python
import numpy as np

# 부스 다섯 곳의 대기 시간을 오전과 점심에 한 번씩 확인
대기시간표 = np.arange(10, 60, 5).reshape(2, 5)
print(대기시간표)
```

```python
전체대기시간분 = np.sum(대기시간표)
print('기록된 대기 시간의 합:', 전체대기시간분, '분')
```

### 합계와 평균은 언제 필요할까요?

`합계`는 세 번 확인한 방문객을 모두 더한 수이고, `평균`은 한 번 확인할 때 방문객이 대략 몇 명이었는지 보여 줍니다. 여러 기록을 한 숫자로 요약해 구역끼리 비교할 때 사용합니다.

### 왜 여기서 둘 다 구할까요?

운영본부는 전체 방문이 많은 구역과 한 번 확인할 때 보통 사람이 많은 구역을 함께 알고 싶어 합니다. 세 구역을 똑같이 세 번씩 확인했으므로 이번 자료에서는 합계가 큰 구역이 평균도 큽니다.

| 2차 | 55 | 44 | 72 |
| 3차 | 38 | 50 | 65 |

### 한 번의 큰 수만 보면 왜 위험할까요?

공연 구역의 72명처럼 한 번의 큰 기록만 보면 나머지 두 번을 놓칩니다. 세 번을 모두 더하고 같은 횟수로 나눈 평균을 보면 우연히 한 번 붐빈 장면에만 끌려가지 않고 전체 기록을 비교할 수 있습니다.

### 결과는 어디까지 말할 수 있을까요?

이번 결론은 세 구역을 각각 세 번 확인한 연습 자료 안에서만 맞습니다. `np.sum()`은 합계, `np.mean()`은 평균을 구합니다. `round(평균, 2)`는 평균을 소수점 이하 둘째 자리까지 반올림합니다. 아래 코드의 평균 출력은 먹거리·체험·공연 순서로 **`45.0 41.33 65.67`**이며, `45.0`은 `45.00`과 같은 값입니다. `공연구역방문객`처럼 장면이 보이는 변수명을 쓰면 코드도 방송 원고처럼 읽을 수 있습니다.


## 1-2. 축제 구역 방문객 데이터 분석 예시(NumPy 위주의 실습)

```python
# 세 번의 취재에서 확인한 축제 구역별 방문객 수
먹거리구역방문객 = np.array([42, 55, 38])
체험구역방문객 = np.array([30, 44, 50])
공연구역방문객 = np.array([60, 72, 65])

먹거리방문객합계 = np.sum(먹거리구역방문객)
체험방문객합계 = np.sum(체험구역방문객)
공연방문객합계 = np.sum(공연구역방문객)

먹거리방문객평균 = np.mean(먹거리구역방문객)
체험방문객평균 = np.mean(체험구역방문객)
공연방문객평균 = np.mean(공연구역방문객)

가장큰방문객합계 = max(먹거리방문객합계, 체험방문객합계, 공연방문객합계)
가장작은방문객합계 = min(먹거리방문객합계, 체험방문객합계, 공연방문객합계)

print('구역별 합계:', 먹거리방문객합계, 체험방문객합계, 공연방문객합계)
print('구역별 평균:', round(먹거리방문객평균, 2), round(체험방문객평균, 2), round(공연방문객평균, 2))
print('가장 큰 합계:', 가장큰방문객합계)
print('가장 작은 합계:', 가장작은방문객합계)
```

구역별 혼잡을 계산한 뒤, 학생들이 축제에서 본 일을 방송부에 알려 준 소식이 시간대별 숫자로 도착했습니다. 이런 소식을 **제보**라고 합니다. 제보 수는 시간대마다 들어온 축제 소식의 개수입니다. 이 수를 보면 언제 소식이 몰렸는지 알 수 있어 방송부가 바쁜 시간의 소식을 먼저 정리할 수 있습니다.

### Series는 언제 필요할까요?

`Series`는 **이름표마다 값 하나**를 짝지어 기록할 때 알맞습니다. 요일마다 걸음 수 하나, 과목마다 점수 하나를 적는 경우가 생활 속 예입니다. Series는 이름표 하나와 값 하나를 한 쌍으로 묶어 한 줄로 정리합니다.

### 왜 여기서 Series를 쓸까요?

방송부는 소식이 언제 몰렸는지 알아야 바쁜 시간의 내용을 먼저 정리할 수 있습니다. 시간대마다 확인할 값은 **축제 소식 수 한 가지**이므로, 각 숫자에 `1교시`, `점심` 같은 시간 이름표를 붙이는 Series가 알맞습니다.

### 숫자 배열만 두면 왜 불편할까요?

NumPy 배열에 `12, 20, 35, 18, 9`만 있으면 `35`가 어느 시간의 값인지 바로 알기 어렵습니다. 세 번째가 점심이라고 따로 외워야 하기 때문입니다. Series로 바꾸면 `점심 → 35건`처럼 숫자의 뜻을 이름으로 바로 확인할 수 있습니다.

> 한 부스에 번호·이름·대기 시간처럼 여러 값을 함께 기록할 때는 다음 단계의 `DataFrame`이 더 알맞습니다.

### 이름표와 자리는 어떻게 다를까요?

`loc`는 **이름표(label)**로 찾고, `iloc`는 **정수 자리(integer position)**로 찾습니다. `label location`과 `integer location`은 이 차이를 떠올리는 기억법이며 공식 풀네임의 정의가 아닙니다. 시험에서는 영문 이름 암기보다 **이름표로 찾는지, 0부터 세는 자리로 찾는지**를 판단합니다.

| 찾는 코드 | 찾는 기준 | 이번 결과 |
|---|---|---:|
| `loc['점심']` | 점심이라는 이름표 | 35 |
| `iloc[2]` | 0부터 세어 세 번째 자리 | 35 |

- `Series`: 시간대마다 축제 소식 수 하나를 짝지은 한 줄 자료
- `loc['점심']`: **점심이라는 이름표**로 찾기
- `iloc[2]`: **0부터 세어 세 번째 자리**로 찾기

점심에 들어온 축제 소식은 35건입니다. `점심`은 사람이 읽는 이름이고, `2`는 컴퓨터가 세는 자리 번호이므로 서로 다른 길로 같은 35건에 도착할 수 있습니다.


## 2-1. Pandas 위주의 실습(이론)




```python
import numpy as np
import pandas as pd

시간대별축제소식수배열 = np.array([12, 20, 35, 18, 9])
시간대별축제소식수배열
```

```python
# 언제 소식이 몰렸는지 알아야 바쁜 시간의 내용을 먼저 정리할 수 있습니다.
# 시간대마다 값이 하나이므로 숫자에 시간 이름표를 붙인 Series를 만듭니다.
시간대별축제소식수 = pd.Series(
    시간대별축제소식수배열,
    index=['1교시', '2교시', '점심', '5교시', '방과후']
)
print(시간대별축제소식수)
```

## 제보 4. 축제 부스표에서 필요한 칸을 찾다

>
> **확보한 단서:** 운영본부가 다섯 부스의 번호·이름·대기 시간·만족도를 보냈습니다.
>
> **이번 취재 임무:** DataFrame으로 묶고 필요한 행·열·한 칸을 정확히 찾습니다.

> **다음 연결:** 전체 표를 읽은 뒤 지원 기준에 맞는 부스만 고릅니다.

점심 소식 35건을 이름표로 찾자 운영본부가 다섯 부스의 번호·이름·대기 시간·만족도를 한 표로 보냈습니다. 이제 시간대 한 값이 아니라 부스마다 여러 항목을 함께 보아야 합니다.

### DataFrame은 언제 필요할까요?

`DataFrame`은 학생마다 여러 과목 점수를 적는 성적표처럼, **같은 종류의 대상이 여러 개**이고 **대상마다 함께 살펴볼 항목도 여러 개**일 때 알맞습니다. 행 하나에는 대상 하나를, 열마다 확인할 항목 하나를 놓습니다.

### 왜 여기서 DataFrame을 쓸까요?

운영본부의 질문은 **“어느 부스를 먼저 도울까?”**입니다. 이 질문에는 부스 이름만으로 답할 수 없습니다. 같은 부스의 `대기시간분`과 `만족도`를 함께 보아야 합니다. 그래서 행 하나를 **부스 하나**로 정하고, 그 부스의 `부스번호`, `부스이름`, `대기시간분`, `만족도`를 옆 칸에 나란히 놓았습니다. 이렇게 하면 한 부스의 정보를 한눈에 확인하고 부스끼리 같은 기준으로 비교할 수 있습니다.

### Series 여러 줄로 두면 왜 불편할까요?

앞 단계의 `Series`는 시간대마다 축제 소식 수 **한 가지**를 찾는 데 알맞았습니다. 이번에는 부스마다 네 가지 정보가 있습니다. Series 여러 개에 따로 두면 같은 부스의 값이 흩어져 맞춰 보기가 불편합니다. DataFrame 한 표에 모으면 부스 한 줄만 보아도 네 정보를 함께 확인할 수 있고, 다음 단계에서 조건에 맞는 부스도 바로 고를 수 있습니다.

| 비교 | 앞 장면의 Series | 이번 장면의 DataFrame |
|---|---|---|
| 한 대상에 붙는 값 | 시간대마다 소식 수 1개 | 부스마다 번호·이름·대기 시간·만족도 4개 |
| 이번에 필요한 일 | 점심의 소식 수 찾기 | 어느 부스를 도울지 여러 항목으로 비교하기 |

> **loc와 iloc 비교:** 지금 표의 행 이름표가 `0, 1, 2`처럼 자리 번호와 같아서 `loc[2]`와 `iloc[2]`는 모두 미니게임 행을 찾습니다. 하지만 `loc[2]`는 **2라는 행 이름표**, `iloc[2]`는 **0부터 세어 세 번째 자리**로 찾습니다. 결과가 같아도 찾는 기준은 다릅니다.

> **결과가 달라지는 예:** 행 이름표가 `2, 5, 8`인 작은 표에서는 `loc[2]`가 이름표 2인 첫째 행 **먹거리**를 찾고, `iloc[2]`가 0부터 세어 세 번째 행 **미니게임**을 찾습니다.

- `축제부스표.loc[2]`: 2라는 행 이름표로 미니게임 부스 찾기
- `축제부스표.iloc[2]`: 0부터 세어 세 번째 자리의 미니게임 부스 찾기
- `축제부스표.loc[3, '대기시간분']`: ① 행 이름표 3인 **음료**를 찾고 ② **대기시간분** 열을 찾기 → 만나는 값은 **14분**
- `축제부스표.loc[:, '부스이름']`: 행 자리의 콜론(`:`)은 **모든 행**이라는 뜻 → 다섯 부스의 이름만 꺼내기
- `축제부스표.loc[:2, :'대기시간분']`: 행은 처음부터 이름표 2까지, 열은 처음부터 대기시간분까지 고르기 → 끝의 2와 대기시간분도 포함

### 표는 어떻게 읽을까요?

1. 쉼표 **왼쪽**에서 행을 찾습니다.
2. 쉼표 **오른쪽**에서 열을 찾습니다.
3. 콜론 하나(`:`)는 그쪽을 **처음부터 끝까지 모두** 고른다는 뜻입니다. `:2`는 처음부터 **이름표 2까지** 고릅니다.
4. `loc`로 범위를 고르면 **마지막 이름표도 포함**합니다.

`iloc`도 쉼표 순서는 같지만 이름표가 아니라 0부터 세는 자리를 사용합니다.


## 2-2. Pandas 실전 예제

```python
import pandas as pd

# 어느 부스를 먼저 도울지 판단하려면 이름·대기 시간·만족도를 함께 보아야 합니다.
# 부스 하나의 정보를 같은 행에 모으기 위해 DataFrame을 만듭니다.
축제부스표 = pd.DataFrame({
    '부스번호': [101, 102, 103, 104, 105],
    '부스이름': ['먹거리', '사진관', '미니게임', '음료', '책갈피'],
    '대기시간분': [18, 7, 25, 14, 9],
    '만족도': [4.2, 4.7, 4.4, 4.1, 4.6]
})

축제부스표
```

```python
# 지금 표에서는 결과가 같지만 찾는 기준은 다릅니다.
print('loc[2]: 2라는 행 이름표로 찾기')
print(축제부스표.loc[2])
print('\niloc[2]: 0부터 세어 세 번째 자리로 찾기')
print(축제부스표.iloc[2])

# 행 이름표가 2, 5, 8이면 loc[2]와 iloc[2]의 결과도 달라집니다.
비교용부스표 = pd.DataFrame(
    {'부스이름': ['먹거리', '사진관', '미니게임']},
    index=[2, 5, 8]
)
비교용부스표.index.name = '행이름표'
print('\n행 이름표가 2, 5, 8인 비교표')
print(비교용부스표)
print('\nloc[2] 결과:', 비교용부스표.loc[2, '부스이름'])
print('iloc[2] 결과:', 비교용부스표.iloc[2]['부스이름'])
```

```python
# ① 쉼표 왼쪽: 행 이름표 3인 음료 부스를 찾습니다.
# ② 쉼표 오른쪽: 대기시간분 열을 찾습니다.
음료부스대기시간 = 축제부스표.loc[3, '대기시간분']
print('음료 부스의 대기 시간:', 음료부스대기시간, '분')
```

```python
# 행 자리의 :는 첫 행부터 마지막 행까지 모두라는 뜻입니다.
모든부스이름 = 축제부스표.loc[:, '부스이름']
print('다섯 부스의 이름')
print(모든부스이름)
```

```python
축제부스표.loc[:, :]
```

```python
# 행: 처음부터 이름표 2까지 고릅니다.
# 열: 처음부터 대기시간분까지 고릅니다.
# loc의 범위에서는 끝 이름표 2와 끝 열 대기시간분도 포함합니다.
처음세부스기본정보 = 축제부스표.loc[:2, :'대기시간분']
처음세부스기본정보
```

### 조건 선택은 언제 필요할까요?

부스가 많을 때 “대기 시간이 15분을 초과한 곳”처럼 기준에 맞는 행만 골라야 할 때 사용합니다. 비교식은 각 부스가 기준에 맞으면 `True`, 아니면 `False`인 판정표를 만듭니다. 그 판정표를 `loc` 안에 넣으면 `True`인 부스만 남습니다.

> **코드를 세 걸음으로 읽기**
> 1. 비교식으로 부스마다 **맞음(`True`)·아님(`False`) 스티커**를 붙입니다.
> 2. `&`는 두 스티커가 **모두 True**, `|`는 **하나 이상 True**인지 확인합니다.
> 3. `축제부스표.loc[조건]`이 True 스티커가 붙은 부스만 남깁니다.

### 코드를 한 줄씩 아주 쉽게 읽어 볼까요?

- `축제부스표['대기시간분'] > 15`: 각 부스의 대기 시간을 보며 **15분보다 긴가?**라고 묻습니다. 15분은 포함하지 않습니다.
- `대기시간15분초과조건`: 질문의 답인 True·False 목록에 붙인 쉬운 이름입니다.
- `축제부스표.loc[대기시간15분초과조건]`: True라고 답한 부스만 표에 남깁니다.
- `축제부스표['만족도'] >= 4.5`: 만족도가 **4.5점 이상인가?**라고 묻습니다. 4.5점도 포함합니다.
- `&`: 왼쪽 조건과 오른쪽 조건을 **모두** 만족해야 남깁니다.
- `|`: 두 조건 중 **하나 이상** 만족하면 남깁니다. 두 조건을 모두 만족한 부스도 포함합니다.

> **꼭 구분하기:** `> 15`는 15분을 빼는 **초과**, `>= 15`는 15분도 넣는 **이상**입니다. 코드마다 만족도 기준도 4.5, 4.4, 4.6으로 다르므로 숫자를 함께 읽습니다.

### 왜 여기서 부스를 걸러낼까요?

운영본부에는 전체 부스표보다 지금 안내하거나 도울 **후보 목록**이 필요합니다. 대기 시간과 만족도 기준을 적용하면 지원이 필요한 곳에 집중할 수 있습니다.

### 눈으로 한 줄씩 고르면 왜 불편할까요?

사람이 긴 표를 눈으로 고르면 15분을 포함할지 빠뜨리기 쉽고, 두 조건을 모두 만족하는지 헷갈릴 수 있습니다. 조건식을 코드로 남기면 모든 부스에 같은 기준을 적용하고 다시 확인할 수 있습니다.

### 조건과 우선순위는 어떻게 읽을까요?

| 코드 | 쉬운 뜻 |
|---|---|
| `대기시간분 > 15` | 15분은 빼고 더 긴 부스 |
| `만족도 >= 4.5` | 4.5점도 포함하는 부스 |
| `&` | 두 조건을 모두 만족 |
| `\|` | 두 조건 중 적어도 하나 만족하며, 둘 다 참이어도 포함 |
| `sort_values(..., ascending=False)` | 큰 값부터 놓아 지원 순서 정하기 |
| `.T` | 값은 그대로 두고 행과 열 바꾸기 |

여러 조건을 연결할 때는 각 비교식을 괄호로 감쌉니다. 정렬은 값을 바꾸는 일이 아니라 행의 순서를 바꾸는 일이고, `.T`는 순위를 정하지 않고 읽는 방향만 바꿉니다.

**두 표를 구분하세요:** 아래 오름차순·내림차순 코드는 `축제부스표`의 **전체 5개 부스**를 정렬합니다. 이어지는 `.T`는 그 정렬표가 아니라 `한조건이상부스`의 **후보 4개 부스**를 사용합니다. 후보는 먹거리·사진관·미니게임·책갈피이며, 전치 결과의 열 이름표는 `0, 1, 2, 4`입니다. 후보만 정렬한 결과나 전체 5부스를 전치한 결과와 섞지 않습니다.

### 추가와 삭제는 어떻게 구분할까요?

대괄호에 없는 열 이름을 지정해 값을 넣으면 맨 뒤에 새 열을 만들고, 기존 열 이름이면 그 열의 값을 덮어씁니다. `insert`는 원하는 자리에 새 열을 끼웁니다. `del`은 한 열을 지우고 `drop`은 이름을 골라 행이나 열을 제외합니다. 합계 열은 삭제 명령을 연습하기 위해 잠시 지웁니다.

`del`과 `insert`는 원래 표를 직접 바꾸고, 여기서 쓰는 `drop`은 원래 표를 유지한 채 제외한 결과를 새 표로 반환합니다. `drop`에서 `axis=0`은 행을 제외하고 `axis=1`은 열을 제외합니다.

### 빈칸 확인은 언제 필요할까요?

표에 `NaN`이 보이거나 기록 개수가 예상보다 적을 때는 분석 전에 빈칸의 위치와 이유를 확인해야 합니다. `NaN`은 결측값, 즉 값이 비어 있거나 알 수 없음을 나타내며 0과 다릅니다. 이 자료에서는 기기 오류나 조사 누락 등으로 기록되지 않은 값을 뜻할 수 있습니다.

### 왜 여기서 바로 0으로 채우지 않을까요?

미니게임의 점심 칸이 빈 이유는 방문객이 없어서일 수도 있지만 조사 누락이나 기기 오류일 수도 있습니다. 이유를 모른 채 0으로 채우면 “모르는 값”을 “아무도 오지 않음”으로 바꾸어 잘못 방송할 수 있습니다.

### 빈칸을 그대로 두면 왜 곤란할까요?

빈칸이 있으면 평균에서 해당 값이 빠지거나 그래프에 점이 나타나지 않아 결과가 달라질 수 있습니다. 그렇다고 무조건 지우면 부스 하나나 시간대 하나의 정보가 함께 사라지므로 처리 전후의 부스 수와 평균을 비교해야 합니다.

### 어떤 처리 방법을 고를까요?

1. 왜 기록되지 않았는지 확인합니다.
2. 부스 행을 지울지, 시간대 열을 지울지, 근거 있는 값으로 채울지 선택합니다.
3. 처리 전후의 부스 수와 평균이 어떻게 달라졌는지 확인합니다.

`dropna`에서 `axis=0`은 빈칸이 있는 행을 제외하고 `axis=1`은 빈칸이 있는 열을 제외합니다. `fillna`는 선택한 값으로 채웁니다. `fillna('확인 중')`은 화면 표시를 보는 기능 확인용 사례입니다. 평균 계산에는 숫자가 필요하므로 나머지 부스의 점심 평균을 대체값으로도 사용하되, 실제 관찰값이 아닌 대체값임을 기록해야 합니다.


```python
# 낮평균을 제외한 복사본으로 누락 처리 연습을 준비합니다.
방문수실수표 = 부스방문표.drop(columns=['낮평균']).astype('float').copy()
방문수실수표
```

```python
# 복사본에 연습용 누락 상황을 만듭니다.
방문수실수표.loc['미니게임', '점심'] = np.nan
방문수실수표
```

```python
빈칸행제외표 = 방문수실수표.dropna(axis=0)
빈칸행제외표
```

```python
빈칸열제외표 = 방문수실수표.dropna(axis=1)
빈칸열제외표
```

```python
문자열로빈칸채운표 = 방문수실수표.astype('object').fillna('확인 중')
문자열로빈칸채운표
```

```python
# 미니게임의 빈 점심 기록을 나머지 부스의 점심 평균으로 채우기
점심평균방문객 = 방문수실수표['점심'].mean()
평균으로빈칸채운표 = 방문수실수표.fillna({'점심': 점심평균방문객})
평균으로빈칸채운표
```

### 그래프 4. 오전·오후 방문객과 대기 시간 분포

첫 코드는 같은 부스 축에서 오전·오후 방문객을 나란한 막대로 비교합니다. 이어지는 상자그림은 축제 중 여러 번 측정한 대기 시간의 중앙값·사분위 범위·이상치를 확인하게 해 줍니다.

> **쉽게 구분하기:** 나란한 막대그래프는 같은 부스의 두 시간대를 비교하고, 상자그림은 여러 대기 시간의 중앙값·사분위 범위·이상치를 확인합니다.

### 상자그림은 어떻게 읽을까요?

대기 시간 메모 10장을 짧은 순서로 늘어놓았다고 생각해 봅시다. **중앙값**은 가운데를 나타내는 값입니다. 자료 수가 짝수인 이번에는 가운데 두 값 12와 14의 평균인 **13분**입니다. **상자**는 가운데 절반 정도의 기록이 놓인 구간을 보여 줍니다. 상자의 왼쪽 끝은 제1사분위수 `Q1`, 오른쪽 끝은 제3사분위수 `Q3`이고, 둘의 차이가 사분위 범위 `IQR`입니다.

**수염**은 상자 밖으로 뻗은 선입니다. 기본 설정에서는 `Q1 - 1.5 × IQR`부터 `Q3 + 1.5 × IQR`까지의 범위 안에 있는 실제 관찰값 중 양끝까지 뻗습니다. 이 범위를 벗어난 관찰값은 점으로 따로 표시하는 **이상치**입니다. 이상치는 드문 값이라는 신호이지, 무조건 틀려서 지워야 하는 값은 아닙니다.

### 이 코드에서는 어떤 값이 나오나요?

대기 시간의 정렬 결과는 `7, 8, 9, 11, 12, 14, 16, 18, 21, 25`입니다. 이 실습의 Matplotlib 기본 상자그림은 NumPy의 **선형 보간** 방식으로 사분위수를 구합니다. 이는 두 값 사이에 필요한 비율만큼 들어간 위치를 계산하는 방법입니다.

| 확인할 값 | 이번 실행 결과 |
|---|---|
| 제1사분위수 `Q1` | 9.5분 |
| 중앙값 `Q2` | 13분 |
| 제3사분위수 `Q3` | 17.5분 |
| 사분위 범위 `IQR` | 17.5 - 9.5 = 8분 |
| 이상치 판단 경계 | 9.5 - 1.5 × 8 = -2.5분, 17.5 + 1.5 × 8 = 29.5분 |
| 수염의 양끝 | 실제 관찰값인 7분과 25분 |
| 이상치 | 없음: 모든 관찰값이 두 경계 사이에 있음 |

경계인 -2.5분과 29.5분은 계산한 기준선이지 실제 대기 시간이나 수염의 끝값이 아닙니다. 교과서의 손계산 규칙과 선형 보간 규칙은 사분위수가 다를 수 있으므로 **시험에서는 어떤 계산법 또는 어떤 코드의 실행 결과를 따르는지 명시합니다.**

<details><summary><b>선형 보간 계산 확인하기</b></summary>

1. 정렬된 10개 값의 자리를 0부터 9까지 셉니다.
2. 25% 위치는 `(10 - 1) × 0.25 = 2.25`입니다. 2번 자리의 9에서 3번 자리의 11 쪽으로 0.25만큼 이동하므로 `Q1 = 9 + (11 - 9) × 0.25 = 9.5`입니다.
3. 50% 위치는 `4.5`입니다. 4번 자리의 12와 5번 자리의 14 사이 가운데이므로 `Q2 = 13`입니다.
4. 75% 위치는 `6.75`입니다. 6번 자리의 16에서 7번 자리의 18 쪽으로 0.75만큼 이동하므로 `Q3 = 16 + (18 - 16) × 0.75 = 17.5`입니다.

</details>

## 마지막 방송. 가장 긴 줄을 근거 있게 알리다

> **이야기 위치:** 3막 · 방송 화면과 보도 완성

> **확보한 단서:** 모든 계산과 표, 그래프가 준비되었지만 숫자만 읽으면 방송이 완성되지 않습니다.
>
> **이번 취재 임무:** 사실·숫자 근거·조사 한계를 묶어 운영본부의 첫 질문에 답합니다.
>
> **방송 다음:** 더 많은 날짜와 시간의 자료를 모으면 결론의 범위를 넓힐 수 있습니다.

마지막 파일에는 다섯 부스의 대기 시간과 만족도가 들어 있습니다. 이제 운영본부가 처음 보낸 “어디가 붐비고 어디를 도와야 할까요?”에 답할 차례입니다. 분석은 가장 큰 숫자를 찾는 데서 끝나지 않습니다.

### 분석 문장은 언제 완성될까요?

코드가 숫자를 출력했을 때가 아니라, 그 숫자가 운영본부의 질문에 어떤 답을 주는지 설명했을 때 분석이 완성됩니다. 표와 그래프에서 직접 본 사실을 말하고 숫자로 근거를 붙여야 합니다.

### 왜 사실·근거·한계를 함께 말할까요?

사실만 말하면 왜 믿어야 하는지 부족하고, 숫자만 말하면 그 수가 무엇을 뜻하는지 알기 어렵습니다. 조사한 범위의 한계까지 밝혀야 듣는 사람이 결론을 어디까지 믿고 사용할지 판단할 수 있습니다.

> **방송 문장 = 눈으로 확인한 사실 + 숫자 근거 + 조사 범위의 한계**

### 가장 큰 숫자만 말하면 왜 위험할까요?

미니게임의 대기 시간이 이번 한 번 25분으로 가장 길어도 언제나 가장 붐비거나 운영이 나쁘다고 단정할 수 없습니다. 날짜·시간·날씨·행사 내용이 달라지면 결과도 달라질 수 있습니다.

### 마지막 방송은 어떻게 쓸까요?

“이번에 확인한 다섯 부스 중 미니게임의 대기 시간이 25분으로 가장 길었습니다”처럼 범위를 먼저 밝힙니다. 이어서 숫자 근거를 말하고, 한 번의 축제 기록이라 다른 날까지 일반화할 수 없다는 한계를 덧붙입니다. 방송이 끝나도 한계를 남기는 것이 다음 취재의 출발점입니다.


# 4. 마지막 방송: 근거 있는 분석 완성하기

### 실전 미션 준비

아래 빈 코드 셀은 **내 예상과 질문을 적는 취재 수첩**입니다. 파이썬 주석처럼 줄 앞에 `#`을 붙이면 실행해도 오류가 나지 않습니다.

1. 표를 보고 대기 시간이 가장 긴 부스를 먼저 예상합니다.
2. 만족도가 가장 높은 부스도 눈으로 찾습니다.
3. 다음 분석 코드를 실행해 예상과 결과를 비교합니다.
4. 마지막 문장을 **사실 + 숫자 근거 + 말할 수 없는 범위**로 완성합니다.


```python
import pandas as pd

최종부스분석표 = pd.DataFrame({
    '부스이름': ['먹거리', '사진관', '미니게임', '음료', '책갈피'],
    '대기시간분': [18, 7, 25, 14, 9],
    '만족도': [4.2, 4.7, 4.4, 4.1, 4.6]
})

평균대기시간 = 최종부스분석표['대기시간분'].mean()
평균만족도 = 최종부스분석표['만족도'].mean()
가장긴대기부스 = 최종부스분석표.loc[
    최종부스분석표['대기시간분'].idxmax(), '부스이름'
]
만족도최고부스 = 최종부스분석표.loc[
    최종부스분석표['만족도'].idxmax(), '부스이름'
]

print(f'평균 대기 시간: {평균대기시간:.1f}분')
print(f'평균 만족도: {평균만족도:.1f}점')
print(f'가장 긴 줄의 부스: {가장긴대기부스}')
print(f'만족도가 가장 높은 부스: {만족도최고부스}')
```
